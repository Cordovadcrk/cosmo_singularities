```python
import numpy as np
import matplotlib.pyplot as plt

import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))

from src.models import (
    BarotropicPowerDiffusionParams,
    BarotropicPowerDiffusionModel,
)
```

```python
a_values = np.logspace(0, 3, 1000)  # a desde 1 hasta 1000
```

```python
params = BarotropicPowerDiffusionParams(
    omega=-1.2,
    alpha=0.05,
    beta=1.5,
    rho0=1.0,
    a0=1.0,
)

model = BarotropicPowerDiffusionModel(params)

rho_values = model.rho_of_a(a_values)

plt.figure(figsize=(7, 4))
plt.plot(a_values, rho_values)
plt.xscale("log")
plt.yscale("log")
plt.xlabel(r"$a$")
plt.ylabel(r"$\rho(a)$")
plt.title(r"Modelo $P=\omega\rho$, $Q=\alpha\rho^\beta$")
plt.grid(True, which="both", alpha=0.3)
plt.show()
```

```python
def compute_rho_a(
    a_values,
    omega,
    alpha,
    beta,
    rho0=1.0,
    a0=1.0,
):
    params = BarotropicPowerDiffusionParams(
        omega=omega,
        alpha=alpha,
        beta=beta,
        rho0=rho0,
        a0=a0,
    )

    model = BarotropicPowerDiffusionModel(params)
    rho_values = model.rho_of_a(a_values)

    return rho_values
```
